#### recursion for printing numbers 

```python
def pri(n):
    if n==0:
        return
    pri(n-1)
    print(n)
```

```python
pri(5)
```

### Factorial

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    return n*factorial(n-1)
```

```python
factorial(3)
```
