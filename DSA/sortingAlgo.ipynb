## Bubble sort

```python
arr=[5,4,1,9,0]
```

```python
def bubble_sort(arr):
    for i in range(len(arr)):
        for j in range(len(arr)-1-i):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
    return arr
```

```python
bubble_sort(arr)
```

## Selection Sort

```python
def selection_sort(arr):
    for i in range(len(arr)-1):
        small=i
        for j in range(i+1, len(arr)-1):
            if arr[j]<arr[small]:
                small=j
        arr[i],arr[small]=arr[small],arr[i]

    return arr

            
```

```python
selection_sort(arr)
```
