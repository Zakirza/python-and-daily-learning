GREEDY ALGORITHMS



| Problem type | Decision variable | Key rule / constraint | Complexity | Common solving technique |
|---|---|---|---|---|
| 0/1 Knapsack | \(x_i \in \{0,1\}\) | Each item is either taken once or skipped. | NP-complete | Dynamic programming: \(O(nW)\) (pseudo-polynomial), or branch and bound |
| Fractional Knapsack | \(0 \le x_i \le 1\) | Items can be divided into fractional parts. | P (polynomial) | Greedy by value-to-weight ratio: \(O(n \log n)\) |
| Bounded Knapsack (BKP) | \(x_i \in \{0,1,\ldots,c_i\}\) | Item \(i\) has at most \(c_i\) copies. | NP-complete | Binary splitting + 0/1 DP: \(O(nW\log C)\), or monotonic deque DP: \(O(nW)\) |
| Unbounded (Complete) Knapsack (UKP) | \(x_i \in \{0,1,2,\ldots\}\) | Unlimited copies of each item are available. | NP-complete | 1D dynamic programming: \(O(nW)\) |
| Multiple-Choice Knapsack (MCKP) | \(\sum_{j \in G_k} x_{kj} \le 1\) | Items are grouped; choose at most one item per group. | NP-complete | Grouped dynamic programming: \(O(W\sum_k |G_k|)\) |
| Multiple Knapsack (MKP) | \(x_{ij} \in \{0,1\}\), \(\sum_j x_{ij} \le 1\) | There are \(m\) knapsacks with capacities \(W_j\); each item can go in at most one. | Strongly NP-hard | MILP, branch and cut, or dynamic programming |
