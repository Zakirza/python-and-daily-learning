## reverse arr

```python
new_arr=[]
def reversearr(arr):
    new_arr=[]
    for i in range(len(arr)-1,-1,-1):
        new_arr.append(arr[i])

    return new_arr #arr[::-1]  this both arr will not modify the original array 
```

```python
reversearr(arr)
```

```python
def reversearr1(arr):
    n=len(arr)
    for i in range(n//2):
        arr[i],arr[n-1-i]=arr[n-1-i],arr[i]
    return arr
```

```python
reversearr1(arr)
```

Rotate a array by d  

```python
arr1=[10,20,30,40,50,60,70]
```

```python
def rotate_a_array(arr,d):
    n=len(arr)
    while d!=0:
        temp=arr[0]
        for i in range(n-1):
            arr[i]=arr[i+1]
        arr[n-1]=temp
        d-=1
    return arr
```

```python
rotate_a_array(arr1,2)
```

## optimizing array rotation 

```python
def rotatesubarray(arr, start, end):
    diff=end-start
    for i in range(diff//2 + 1):
        arr[start+i],arr[end-i]=arr[end-i],arr[start+i]

def rotate(arr,n,d):
    d=d%n
    rotatesubarray(arr,0,d-1)
    rotatesubarray(arr,d, n-1)
    rotatesubarray(arr,0, n-1)
     
```

```python
rotate(arr1,7,3)
```
