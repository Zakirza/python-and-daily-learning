Stack using list

```python
class Stack:
    def __init__(self):
        self.items=[]
    def push(self,val):
        self.items.append(val)
    def top(self):
        if len(self.items)==0:
            return "cannot find stack is empty"
        return self.items[-1]
    def pop(self):
        if len(self.items)==0:
            return "cannot pop stack is empty"
        x=self.items.pop()
        return x
    def is_empty(self):
        return len(self.items)==0

    def size(self):
        return len(self.items)
```

```python
st=Stack()
st.push(10)
st.push(12)
st.push(19)
```
