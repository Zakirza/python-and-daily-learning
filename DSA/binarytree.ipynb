```python
class Node:
    def __init__(self,val):
        self.val=val
        self.right=None
        self.left=None
```

```python
drinks=Node("drinks")
hot=Node("hot")
cold=Node("cold")
tea=Node("tea")
coffee=Node("coffee")
cola=Node("cola")
fanta=Node("fanta")
```

```python
hot.left=tea
hot.right=coffee

cold.left=cola
cold.right=fanta

drinks.left=hot
drinks.right=cold
```

```python
print(drinks.val)
print(drinks.right.val)
print(cold.val)
print(drinks.left.left.val)
print(drinks.left.right.left.val)
```

## Binary tree traversal DFS AND BFS

```python
one=Node(1)
two=Node(2)
three=Node(3)
four=Node(4)
five=Node(5)
six=Node(6)
seven=Node(7)
eight=Node(8)
nine=Node(9)
ten=Node(10)
```

```python
three.left=two
three.right=nine

eight.left=one
eight.right=six

four.left=eight
four.right=ten

five.left=three
five.right=four
```

```python
def preorder(Node):
    if Node==None:
        return 
    print(Node.val,end=" ")
    preorder(Node.left)
    preorder(Node.right)
```

```python
preorder(five)
```

```python
def inorder(Node):
    if Node==None:
        return 
    inorder(Node.left)
    print(Node.val)
    inorder(Node.right)
```

```python
inorder(five)
```

```python
def postorder(Node):
    if Node==None:
        return
    postorder(Node.left)
    postorder(Node.right)
    print(Node.val)
```

```python
postorder(five)
```

```python
from collections import deque
def level_order(Node):
    result=[]
    queue=deque([Node])
    while len(queue) != 0:
        e=queue.popleft()
        result.append(e.val)
        if e.left != None:
            queue.append(e.left)
        if e.right != None:
            queue.append(e.right)
    return result
```

```python
level_order(five)
```
