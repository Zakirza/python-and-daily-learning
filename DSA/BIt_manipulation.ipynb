### Bit manipulation 

```python
def convert_to_binary(num:int)->str:
    result=""
    while num > 0:
        if num %2==1:
            result+="1"
        else:
            result+="0"
        num=num//2
    return result[::-1]
```

```python
convert_to_binary(13)
```

```python
def convert2decimal(x:str)->int:
    n=len(x)
    reverse=x[::-1]
    c=0
    for i in range(n):
        c+=int(reverse[i])*(2**i)

    return c
```

Swap two numbers

check if ith bit is set or not by using left shift(>>)

```python
def checkithbitset(num:int,i)->bool:
    if num&(1<<i)!=0:
        return 1
    else:
        return 0
```

```python
checkithbitset(13,1)
```

check if ith bit is set or not by using right shift(<<)

```python
def rightshiftithbitcheck(num:int,i:int)->bool:
    if num>>i&1:
        return 1
    else:
        return 0
```

```python
rightshiftithbitcheck(13,2)
```
