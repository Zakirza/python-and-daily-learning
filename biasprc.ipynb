Printing numbers

```python
def print_numbers(n):
    if n==0:
        return
    print_numbers(n-1)
    print(n)

```

```python
print_numbers(4)
```

Factorial

```python
def fac(n):
    if n==0:
        return 1
    return n*fac(n-1)
    
```

```python
fac(4)
```

# subset of list

```python
subset=[]
n=4

def search(k):
    if k == n:
        print(subset)
        return
    search(k+1)
    subset.append(k)
    search(k+1)
    subset.pop()
```

```python
search(0)
```

```python
def f(n):
    if n == 0:
        print("STOP")
        return
    print(n)
    f(n-1)
```

```python
f(6)
```

 sum of n numbers

```python

def sum(n):
    if n==0:
        return 0
    return n+sum(n-1)
```

```python
sum(100)
```

PERMUTATIONS

```python
lst=[1,2,3]
def permutaion(lst,path=[]):
    if not lst:
        print(path)
        return 
    for i in range(len(lst)):
        permutaion(lst[i]+lst[i+1],path+[lst[i]])
```
